```python
### Read the files into dataframes
import pandas as pd
import glob

files = glob.glob("*.csv")

dfs = [pd.read_csv(file) for file in files]
```

```python
ggle = dfs[0]
fb = dfs[1]
amzn = dfs[2]
nflx = dfs[3]
aapl = dfs[4]
```

```python
### c

aapl['ticker'] = 'aapl'
amzn['ticker'] = 'amzn'
fb['ticker'] = 'fb'
nflx['ticker'] = 'nflx'
ggle['ticker'] = 'ggle'

### d
df = pd.concat([aapl, amzn, fb, nflx, ggle])
df
```
