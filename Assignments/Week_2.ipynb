```python
letters = ["a","b","c","d"]
print(letters)

b = letters[int(int('3'*2)//11)]
print(b)
# '3' * 2 is '33'
# int('33') is 33
# 33 // 11 is 3
# letters[3] is the fourth item on the list: "d"

c = letters[-1]
print(c)
# index -1 means the last element in the list: 'd'

d = letters[:2]
print(d)
# it takes all the item before the thrid one, so only "a" and "b"
```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']

def syntax(spam):
  if len(spam) == 0:
    return ''
  elif len(spam) == 1:
    return spam[0]
  else:
    return ', '.join(spam[:-1]) + ', and ' + spam[-1]

print(syntax(spam))
```
