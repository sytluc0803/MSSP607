```python
from pathlib import Path
import pandas as pd

tickers = ["fb", "aapl", "amzn", "nflx", "goog"]

dfs = []
for t in tickers:
    temp = pd.read_csv(f"{t}.csv")
    temp["ticker"] = t
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)
```

```python
df = (
    pd.read_csv("faang.csv", parse_dates=["date"])
      .astype({"volume": "int64"})
      .sort_values(["date", "ticker"])
      .reset_index(drop=True)
)
df
```

```python
#change sort_values(...).head(7) to nlargest()

top7 = df.nlargest(7, "volume")
top7
```

```python
#Naming the output columns and sorting the result for efficiency

# original: df_melted = df.melt(id_vars=["date", "ticker"])

df_melted = df.melt(
    id_vars=["date", "ticker"],
    var_name="metric",
    value_name="value",
).sort_values(["date", "ticker", "metric"]).reset_index(drop=True)

df_melted
```
