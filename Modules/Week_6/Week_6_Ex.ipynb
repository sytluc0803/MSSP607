```python
import pandas as pd
df = pd.read_csv('Student Performance Data.csv')
df.head()
```

```python
# add a score_category that assign different score range to different grades
df["score_category"] = df["total score"].apply(
    lambda x: "Excellent" if x >= 270
    else "Good" if x >= 230
    else "Average" if x >= 180
    else "Needs improvement"
)

df.head()
```
