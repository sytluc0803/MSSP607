```python
import pandas as pd
df = pd.read_csv('InternetSales.csv', encoding='latin1')
df.head()
```

```python
df.shape
```

```python

# Rename Variable
df = df.rename(columns={'EnglishCountryRegionName': "CountryName",
                        "SalesAmount":"Sales",
                        "EnglishProductName": "ProductName"})
print(df["CountryName"].unique())
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert OrderDate to datetime type
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

# Create a new column called Year
df["OrderYear"] = df["OrderDate"].dt.year
df.head()
```

# What are the Internet sales of Adventure Works by year for each Country Region?

```python
print(df["OrderYear"].unique())
print(df["CountryName"].unique())
```

```python
yearly_sales = df.groupby("OrderYear")["Sales"].sum()
yearly_sales
```

Sales of Year 2010 and 2014 have unusual discrepency with the other years, a maginitude smaller by 100 and 1000 times. So it is worth examine the sales by quarters instead of years.

```python
result = df.groupby([
    df["OrderDate"].dt.year.rename("Year"),
    df["OrderDate"].dt.quarter.rename("Quarter")
])["Sales"].sum()

result
```

Given that it is still a 100 times size difference on quarter scale, data from 2010 and 2014 could be not contain a complete quarter, so the following analysis will focus on year 2011 to year 2013.

```python
df1 = df[~df["OrderDate"].dt.year.isin([2010, 2014])]
df1.head()
```

```python
print(df1["OrderYear"].unique())
```

```python

Country_yearly_sales = df1.groupby(["CountryName", "OrderYear"])["Sales"].sum()
print(Country_yearly_sales)

# Reshape: countries as rows, years as columns
sales_pivot = Country_yearly_sales.unstack()

# Create grouped bar chart
sales_pivot.plot(kind="bar", figsize=(12, 6))

plt.title("Yearly Sales by Country")
plt.xlabel("Country")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Year")
plt.tight_layout()
plt.show()
```

Australia and United States are the bellcows of the internet sales company,they have more sales than other countries all three years. In the particularly better year 2013, US has 5 million dollar sales when countries other than AU barely top 1 million.

```python
summary = df1.groupby(
    ["CountryName"]
)["Sales"].describe().round(2)

print(summary)

df1.boxplot(column="Sales", by="CountryName", figsize=(12, 6))

plt.title("Sales Distribution by Country")
plt.suptitle("")  # Remove the default title
plt.xlabel("Country")
plt.ylabel("Sales per order")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

```

#What is the average price of the product category that generates the most sales?

```python
print(df1['ProductName'].value_counts())
```

```python
topsales = df1.groupby('ProductName')["Sales"].sum()
print(topsales.nlargest(20))
```

```python
# How many products are there in the Mountain-200 series?""
mountain_products = df1.loc[
    df1["ProductName"].str.contains("Mountain-200", na=False),
    "ProductName"
].unique()

list(mountain_products)
```

Now the lists match, the Mountain-200 series ara the most sold categories with 6 of 6 in global top sales. Now the next step is to calculate the average price

```python
# make sure each row is exactly one order
df1["OrderQuantity"].value_counts()
```

```python
# double check
mountain_prices = df1.loc[
    df1["ProductName"].isin(mountain_products),
    ["ProductName", "UnitPrice"]
]

mountain_prices["UnitPrice"].mean()
```

```python

mountain_products = list(mountain_products)
mountain_sales = df1[df1["ProductName"].isin(mountain_products)]

average_price = mountain_sales["UnitPrice"].sum() / len(mountain_sales)

print(average_price)
```

```python
# What are the top runners in each countries?
# Since there are six product in Mountain 200 Series, we do top six
country_sales = df1.groupby(["CountryName", "ProductName"])["Sales"].sum()
top6_products = country_sales.groupby(level="CountryName", group_keys=False).nlargest(6)
top6_products
```
